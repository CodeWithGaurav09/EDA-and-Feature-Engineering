```python
df_country=pd.read_excel('Country-Code.xlsx')
df_country.head()
```

```python
df.columns
```

```python
final_df=pd.merge(df,df_country,on='Country Code', how='left')
final_df.head(2)
```

```python
final_df.columns
```

```python
## To check data types of the columns in the dataset
final_df.dtypes
```

```python
final_df.Country.value_counts()
```

```python
Country_Name=final_df.Country.value_counts().index
Country_Name
```

```python
Country_Name.values
Country_Val=final_df.Country.value_counts().values
Country_Val
```

```python


## Pie Chart- Top 3 countries that uses zomato
import matplotlib.pyplot as plot
plt.pie(Country_Val[:3],labels=Country_Name[:3],autopct='%1.2f%%')
```

 Observation:In Zomato All the Transaction from the India , after That United State and United Kingdom.

```python
final_df.columns
```

```python
ratings=final_df.groupby(['Aggregate rating','Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
ratings
```

```python
ratings.head(10)
```

Observation: 

1.When Rating is between 4.5 to 4.9---> Excellent
2.When Rating are between 4.0 to 3.4--->very good
3.when Rating is between 3.5 to 3.9----> good
4.when Rating is between 3.0 to 3.4----> average
5.when Rating is between 2.5 to 2.9----> average
6.when Rating is between 2.0 to 2.4----> Poor


```python

import matplotlib
matplotlib.rcParams['figure.figsize'] = (12,6)
sns.barplot(x='Aggregate rating',y='Rating Count',data=ratings,hue='Rating color',palette=['red','orange','yellow','green','green'])
```

Observation:

Not Rated count is very high
Maximum number of rating are between 2.5 to 3.4

```python
## Count plot
sns.countplot(x="Rating color",data=ratings,palette=['blue','red','orange','yellow','green','green'])
```
